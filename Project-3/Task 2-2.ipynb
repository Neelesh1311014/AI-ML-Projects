import libraries

```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
# Load the dataset
df = pd.read_csv("Titanic-Dataset.csv")

print(df.head())
```

```python
print(df.info())

print(df.describe())

print(df.isnull().sum())
```

```python
df.duplicated().sum()
```

handleing missing values

```python
# Fill missing Age with median
df['Age'].fillna(df['Age'].median(), inplace=True)

# Fill missing Embarked with mode
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

# Fill missing Cabin with 'Unknown'
df['Cabin'].fillna('Unknown', inplace=True)

print(df.isnull().sum())
```

Remove Duplicate Records

```python
print("Before removing duplicates:", df.shape)

df.drop_duplicates(inplace=True)

print("After removing duplicates:", df.shape)
```

Correct Data Types

```python
print(df.dtypes)

# Convert Age to integer
df['Age'] = df['Age'].astype(int)

print(df.dtypes)
```

Feature Engineering

```python
#(a) Extract Title from Name
df['Title'] = df['Name'].str.extract(' ([A-Za-z]+)\.', expand=False)

print(df[['Name','Title']].head())

#(b)Create Family Size
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

print(df[['FamilySize']].head())

#(c)Create Age Categories
df['AgeGroup'] = pd.cut(df['Age'],
                        bins=[0,12,19,35,60,100],
                        labels=['Child','Teen','Adult','Middle Age','Senior'])

print(df[['Age','AgeGroup']].head())
```

Convert Categorical Features to Numerical

```python
encoder = LabelEncoder()

df['Sex'] = encoder.fit_transform(df['Sex'])

df['Embarked'] = encoder.fit_transform(df['Embarked'])

df['Title'] = encoder.fit_transform(df['Title'])

df['AgeGroup'] = encoder.fit_transform(df['AgeGroup'])

print(df.head())
```

Feature Scaling

```python
scaler = StandardScaler()

df[['Age','Fare','FamilySize']] = scaler.fit_transform(
    df[['Age','Fare','FamilySize']]
)

print(df[['Age','Fare','FamilySize']].head())
```

Display Final Dataset

```python
print(df.head())

print(df.info())
```

Saving the updated dataset

```python
df.to_csv("Titanic_Cleaned.csv", index=False)

print("Updated dataset saved successfully.")
```

Conclusion

The Titanic dataset was successfully preprocessed through data cleaning and feature engineering. Missing values in the Age, Cabin, and Embarked columns were handled appropriately, duplicate records were removed, and data types were verified for consistency. New features such as Title, FamilySize, and AgeGroup were created to provide additional useful information for analysis. Categorical variables were converted into numerical form using Label Encoding, and numerical features were standardized using StandardScaler. These preprocessing steps improved the quality of the dataset, making it clean, consistent, and ready for machine learning algorithms, which can lead to better model performance and more accurate predictions.
